```python
import os
import csv
import time
import requests as req
from bs4 import BeautifulSoup as bs

def scrape_detik(hal):
    headers = {
        'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/138.0.0.0 Safari/537.36 Edg/138.0.0.0'
        }
    a = 1
    filename = 'berita_detik_kunci_crypto.csv'
    file_exists = os.path.exists(filename)

    for page in range(1, hal + 1):
        url = f'https://www.detik.com/search/searchall?query=crypto&result_type=latest&page={page}'
        print(f"\n--- Scraping Page {page} ---")
        ge = req.get(url, headers=headers).text
        sop = bs(ge, 'lxml')
        articles = sop.find_all('article', class_='list-content__item')

        for article in articles:
            try:
                link_tag = article.find('a')
                link = link_tag['href'] if link_tag else 'N/A'
                title_tag = article.find('h3', class_='media__title')
                headline = title_tag.text.strip() if title_tag else 'N/A'

                #ge_ = req.get(link, headers=headers).text
                #sop_ = bs(ge_, 'lxml')

                detail = bs(req.get(link, headers=headers).text, 'lxml')

                content_div = detail.find('div', class_='detail__body-text') or detail.find('div', class_='detail__body-text itp_bodycontent')
                isi = ' '.join([p.text.strip() for p in content_div.find_all('p')]) if content_div else ''

                date_tag = detail.find('div', class_='detail__date')
                date = date_tag.text.strip() if date_tag else 'N/A'
                date = ' '.join(date.split()[:3]) if date != 'N/A' else date

                print(f'done[{a}] > {headline[:50]}')
                with open(filename, 'a', newline='', encoding='utf-8') as file:
                    wr = csv.writer(file)
                    if a == 1:
                        wr.writerow(['headline', 'date_article', 'link', 'content'])
                        file_exists = True
                    wr.writerow([headline, date, link, isi])
                a += 1
                time.sleep(1)

            except Exception as e:
                print(f"Error at item {a}: {e}")
                continue

# Panggil fungsi
scrape_detik(15)
```

```python
import pandas as pd
import glob
import os

# --- 1. Cari semua file CSV yang relevan ---
# Pola ini akan mencocokkan file yang di unggah tadi.
# Tanda '*' artinya "karakter apa saja".
file_pattern = 'berita_detik_kunci_*.csv'
file_names = glob.glob(file_pattern)

print(f"Ditemukan {len(file_names)} file CSV untuk digabungkan:")
for f in file_names:
    print(f" - {f}")

# --- 2. Baca dan gabungkan file ---
if not file_names:
    print("\n[ERROR] Tidak ada file yang ditemukan! Pastikan Anda sudah mengunggah file CSV ke panel kiri Colab.")
else:
    dataframes = []
    print("\nMemproses file...")
    for file in file_names:
        try:
            # Membaca setiap file CSV
            df = pd.read_csv(file)
            dataframes.append(df)
            # print(f"  -> Berhasil membaca: {file} ({len(df)} baris)")
        except Exception as e:
            print(f"  -> Gagal membaca {file}: {e}")

    if dataframes:
        # Menggabungkan semua dataframe dalam list menjadi satu
        merged_df = pd.concat(dataframes, ignore_index=True)
        total_awal = len(merged_df)
        print(f"\nTotal baris data awal: {total_awal}")

        # --- 3. Pre-processing (Hapus Duplikat) ---
        # Menghapus baris jika memiliki 'link' yang sama.
        # keep='first' berarti menyimpan data pertama yang ditemukan, sisanya dihapus.
        merged_df.drop_duplicates(subset=['link'], keep='first', inplace=True)
        total_akhir = len(merged_df)
        print(f"Dihapus {total_awal - total_akhir} duplikat.")
        print(f"Total baris data bersih: {total_akhir}")

        # --- 4. Simpan hasil ke file baru ---
        output_filename = 'berita_crypto_gabungan_bersih.csv'
        merged_df.to_csv(output_filename, index=False, encoding='utf-8-sig')

        print(f"\n[SUKSES] File gabungan disimpan sebagai: {output_filename}")
        print("Cek panel file di sebelah kiri untuk mengunduh hasilnya.")

        # Menampilkan 5 data teratas sebagai sampel
        print("\nContoh 5 data teratas:")
        display(merged_df.head())

    else:
        print("\nGagal menggabungkan karena tidak ada data yang valid.")
```

### Mengurutkan Tanggal

```python
import pandas as pd
import re

# 1. Load data gabungan
df = pd.read_csv('berita_crypto_gabungan_bersih.csv')

# 2. Peta untuk mengubah nama bulan Indonesia ke angka
MONTH_MAP = {
    'Jan': '01', 'Feb': '02', 'Mar': '03', 'Apr': '04', 'Mei': '05', 'Jun': '06',
    'Jul': '07', 'Agu': '08', 'Sep': '09', 'Okt': '10', 'Nov': '11', 'Des': '12'
}

def normalize_date_indonesia(date_str):
    if not isinstance(date_str, str):
        return pd.NaT # Not a Time (untuk data kosong)

    # Bersihkan spasi ekstra
    date_str = date_str.strip()

    # Coba ganti nama bulan menggunakan map
    for ind, eng in MONTH_MAP.items():
        # Gunakan regex untuk mengganti kata yang cocok (case insensitive)
        date_str = re.sub(r'\b' + ind + r'\b', eng, date_str, flags=re.IGNORECASE)

    try:
        # Parsing ke datetime.
        # dayfirst=True penting agar '02 11 2025' dibaca 2 November, bukan 11 Februari
        return pd.to_datetime(date_str, dayfirst=True)
    except:
        return pd.NaT

# 3. Terapkan normalisasi ke kolom baru 'date_iso'
print("Sedang menormalisasi tanggal...")
df['date_iso'] = df['date_article_clean'].apply(normalize_date_indonesia)

# 4. Cek jika ada tanggal yang gagal diparsing (NaT)
failed_dates = df[df['date_iso'].isna()]['date_article_clean'].unique()
if len(failed_dates) > 0:
    print(f"Peringatan: Ada {len(failed_dates)} format tanggal yang gagal diparsing: {failed_dates[:5]}...")

# 5. Mengurutkan Data
# ascending=False artinya menurun (Terbaru ke Terlama)
# ascending=True artinya menaik (Terlama ke Terbaru)
df_sorted = df.sort_values(by='date_iso', ascending=False)

# 6. Simpan hasil urutan
output_sorted = 'berita_crypto_gabungan_sorted.csv'
# Menyertakan kolom 'date_iso' agar mudah dicek, atau Anda bisa menghapusnya jika mau.
df_sorted.to_csv(output_sorted, index=False, encoding='utf-8-sig')

print("\nData berhasil diurutkan berdasarkan tanggal!")
print("5 Berita Terbaru:")
print(df_sorted[['date_iso', 'headline']].head())
print("\n5 Berita Terlama:")
print(df_sorted[['date_iso', 'headline']].tail())
```

```python
import pandas as pd
from transformers import AutoTokenizer, AutoModelForSequenceClassification
import torch
from collections import Counter

# Load data gabungan
df = pd.read_csv("/content/drive/MyDrive/berita_crypto_final_filtered_2024-2025_8.csv")

# Gunakan kolom 'content' sebagai input
texts = df['content'].fillna('').astype(str)
texts = texts[texts.str.strip() != '']

# Load model dan tokenizer
model_name = "ihsan31415/indo-roBERTa-financial-sentiment"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSequenceClassification.from_pretrained(model_name)
model.eval()

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model.to(device)

label_map = {0: "Positive", 1: "Neutral", 2: "Negative"}

# Fungsi untuk memecah teks panjang menjadi chunk ≤512 token
def chunk_text(text, max_tokens=512):
    tokens = tokenizer.tokenize(text)
    chunks = []
    for i in range(0, len(tokens), max_tokens):
        chunk = tokens[i:i+max_tokens]
        chunk_text = tokenizer.convert_tokens_to_string(chunk)
        chunks.append(chunk_text)
    return chunks

# Fungsi prediksi sentimen untuk teks panjang
def predict_long_text(text):
    chunks = chunk_text(text)
    votes = []
    for chunk in chunks:
        inputs = tokenizer(chunk, return_tensors="pt", truncation=True, padding=True, max_length=512)
        inputs = {k: v.to(device) for k, v in inputs.items()}
        inputs.pop('token_type_ids', None)
        with torch.no_grad():
            outputs = model(**inputs)
            pred_id = torch.argmax(outputs.logits, dim=1).item()
            votes.append(pred_id)
    majority = Counter(votes).most_common(1)[0][0]
    return label_map[majority]

# Prediksi seluruh data
df['sentiment'] = texts.apply(predict_long_text)

# Simpan hasil ke Google Drive
df.to_csv("/content/drive/MyDrive/sentiment_gabungan_berita_detik.csv", index=False)

# Distribusi sentimen keseluruhan
print("Distribusi Sentimen:", Counter(df['sentiment']))
```

###Confusion Matrix Fix

```python
import pandas as pd
from transformers import AutoTokenizer, AutoModelForSequenceClassification
import torch
from collections import Counter

# Load data gabungan
df = pd.read_csv("/content/drive/MyDrive/confussion_matrix.csv")

# Gunakan kolom 'content' sebagai input
texts = df['content'].fillna('').astype(str)
texts = texts[texts.str.strip() != '']

# Load model dan tokenizer
model_name = "ihsan31415/indo-roBERTa-financial-sentiment"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSequenceClassification.from_pretrained(model_name)
model.eval()

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model.to(device)

label_map = {0: "Positive", 1: "Neutral", 2: "Negative"}

# Fungsi untuk memecah teks panjang menjadi chunk ≤512 token
def chunk_text(text, max_tokens=512):
    tokens = tokenizer.tokenize(text)
    chunks = []
    for i in range(0, len(tokens), max_tokens):
        chunk = tokens[i:i+max_tokens]
        chunk_text = tokenizer.convert_tokens_to_string(chunk)
        chunks.append(chunk_text)
    return chunks

# Fungsi prediksi sentimen untuk teks panjang
def predict_long_text(text):
    chunks = chunk_text(text)
    votes = []
    for chunk in chunks:
        inputs = tokenizer(chunk, return_tensors="pt", truncation=True, padding=True, max_length=512)
        inputs = {k: v.to(device) for k, v in inputs.items()}
        inputs.pop('token_type_ids', None)
        with torch.no_grad():
            outputs = model(**inputs)
            pred_id = torch.argmax(outputs.logits, dim=1).item()
            votes.append(pred_id)
    majority = Counter(votes).most_common(1)[0][0]
    return label_map[majority]

# Prediksi seluruh data
df['sentiment'] = texts.apply(predict_long_text)

# Simpan hasil
output_path = "/content/drive/MyDrive/confussion_matrix_fix.csv"
df.to_csv(output_path, index=False)
print(f"Hasil sentimen sudah disimpan ke: {output_path}")

# Distribusi sentimen keseluruhan
print("Distribusi Sentimen:", Counter(df['sentiment']))
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix, classification_report

# 1. Load data
df = pd.read_csv("/content/drive/MyDrive/confussion_matrix_fix_manual.csv")

# 2. Mapping kolom 'sentiment' ke numerik
sentiment_map = {
    "Positive": 1,
    "Neutral": 0,
    "Negative": -1
}
df['sentiment'] = df['sentiment'].map(sentiment_map)

# 3. Ambil label aktual dan prediksi
y_true = df['Label']              # label aktual dari file
y_pred = df['sentiment']    # prediksi hasil mapping

# 4. Hitung confusion matrix
cm = confusion_matrix(y_true, y_pred, labels=[-1, 0, 1])

# 5. Visualisasi confusion matrix
plt.figure(figsize=(6,5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=['Negative (-1)','Neutral (0)','Positive (1)'],
            yticklabels=['Negative (-1)','Neutral (0)','Positive (1)'])
plt.xlabel("Prediksi")
plt.ylabel("Aktual")
plt.title("Confusion Matrix Sentimen")
plt.grid(False)
plt.tight_layout()
plt.show()

# 6. Classification report
print("Classification Report:")
print(classification_report(y_true, y_pred, digits=3, target_names=['Negative','Neutral','Positive']))
```

##Model 1 (model hibrida IndoBERT-BiGRU)

Memanggil semua fitur pada dataset akhir untuk dilakukan pemodelan dan pelatihan model hibrida IndoBERT-BiGRU untuk memprediksi variabel target Yt+1.

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import GRU, Bidirectional, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping
import matplotlib.pyplot as plt
from sklearn.metrics import mean_squared_error

# 1. Baca data
df_raw = pd.read_csv("/content/drive/MyDrive/gabungan_varx_sentimen_roberta.csv")
df_raw['date'] = pd.to_datetime(df_raw['date'])
df_raw = df_raw.sort_values('date').reset_index(drop=True)

# 2. Kolom numerik
num_cols = ['x1','x2','x3','x4','x5']
extra_cols = ['D1','D2','D1_lag','D2_lag']

# 3. Bersihkan agar benar-benar float
for c in num_cols + extra_cols:
    df_raw[c] = df_raw[c].astype(str).str.replace(',', '').astype(float)

# 4. Normalisasi
feature_cols = num_cols + extra_cols
scaler_X = MinMaxScaler()
X_scaled = scaler_X.fit_transform(df_raw[feature_cols].astype(float))

scaler_y = MinMaxScaler()
y_scaled = scaler_y.fit_transform(df_raw['x1'].values.reshape(-1,1))

# 5. Bentuk sequence data (timesteps=30)
timesteps = 30
X_seq, y_seq, y_dates = [], [], []
for i in range(len(X_scaled) - timesteps):
    X_seq.append(X_scaled[i:i+timesteps])
    y_seq.append(y_scaled[i+timesteps])
    y_dates.append(df_raw['date'].iloc[i+timesteps])  # simpan tanggal target

X_seq = np.array(X_seq)
y_seq = np.array(y_seq)
y_dates = np.array(y_dates)

# 6. Train-test split (80% train, 20% test)
split = int(len(X_seq) * 0.8)
X_train, X_test = X_seq[:split], X_seq[split:]
y_train, y_test = y_seq[:split], y_seq[split:]
dates_train, dates_test = y_dates[:split], y_dates[split:]

# 7. Model BiGRU (2 lapisan)
model = Sequential()
model.add(Bidirectional(GRU(128, return_sequences=True), input_shape=(timesteps, X_train.shape[2])))
model.add(Dropout(0.3))
model.add(Bidirectional(GRU(64)))
model.add(Dense(32, activation='relu'))
model.add(Dense(1))
model.compile(optimizer='adam', loss='mape', metrics=['mape'])

# 8. Training dengan early stopping
early_stop = EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True)
history = model.fit(X_train, y_train, epochs=200, batch_size=32,
                    validation_data=(X_test, y_test), verbose=1,
                    callbacks=[early_stop])

# 9. Evaluasi
loss, mape = model.evaluate(X_test, y_test)
print(f"Test MAPE: {mape:.2f}%")

# Hitung prediksi pada test set
y_pred_test = model.predict(X_test)
y_pred_rescaled = scaler_y.inverse_transform(y_pred_test)
y_test_rescaled = scaler_y.inverse_transform(y_test)

# RMSE
rmse = np.sqrt(mean_squared_error(y_test_rescaled, y_pred_rescaled))
print(f"Test RMSE: {rmse:.2f}")

# 10. Prediksi hari berikutnya (gunakan 30 hari terakhir)
last_window = df_raw.iloc[-timesteps:][feature_cols].astype(float)
last_window_scaled = scaler_X.transform(last_window)
last_window_scaled = last_window_scaled.reshape(1, timesteps, X_train.shape[2])

predicted_scaled = model.predict(last_window_scaled)
predicted_x1 = scaler_y.inverse_transform(predicted_scaled)[0][0]

last_date = df_raw['date'].iloc[-1]
forecast_date = last_date + pd.Timedelta(days=1)

print(f"Last known date: {last_date.date()}")
print(f"Forecast date:  {forecast_date.date()}")
print(f"Forecast BTC:    {predicted_x1:,.2f}")

# 11. Visualisasi hasil prediksi vs aktual di test set (pakai tanggal)
y_pred_test = model.predict(X_test)
y_pred_rescaled = scaler_y.inverse_transform(y_pred_test)
y_test_rescaled = scaler_y.inverse_transform(y_test)

plt.figure(figsize=(12,6))
plt.plot(dates_test, y_test_rescaled, label='Actual BTC')
plt.plot(dates_test, y_pred_rescaled, label='Predicted BTC')
plt.title("Forecasting BTC")
plt.xlabel("Date")
plt.ylabel("BTC Price")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

# 12. Visualisasi training loss vs validation loss
plt.figure(figsize=(10,5))
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title("Model Training vs Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("MAPE Loss")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import shap

# 8. SHAP KernelExplainer
# Flatten sequence untuk SHAP (model-agnostic)
def f(X):
    X = X.reshape(X.shape[0], timesteps, X_train.shape[2])
    return model.predict(X).flatten()

# Background sample
bg_size = min(100, len(X_train))
background = X_train[:bg_size].reshape(bg_size, -1)

# Test sample
test_size = min(50, len(X_test))
X_test_sample = X_test[:test_size].reshape(test_size, -1)

explainer = shap.KernelExplainer(f, background)
shap_values = explainer.shap_values(X_test_sample)

# 9. Visualisasi SHAP summary
feature_names = []
for t in range(timesteps):
    feature_names += [f"{fname}_t-{timesteps-1-t}" for fname in feature_cols]

shap.summary_plot(shap_values, X_test_sample, feature_names=feature_names)
```
